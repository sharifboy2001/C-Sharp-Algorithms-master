```python
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum as spark_sum
import time
```

```python
def pandas_example(data_size):
    # Ma'lumotlar yaratish
    df = pd.DataFrame({'number': range(data_size)})

    start_time = time.time()
    # Hisoblash: har bir raqamni kvadratga ko'tarish
    result = df['number'].apply(lambda x: x ** 2)
    sum_result = result.sum()
    end_time = time.time()

    print(f"Pandas vaqti: {end_time - start_time:.4f} soniya")
    return sum_result
```

```python
def pyspark_example(data_size):
    # SparkSession yaratish
    spark = SparkSession.builder.appName("PySpark_vs_Pandas").getOrCreate()

    # Ma'lumotlar yaratish
    data = [(i,) for i in range(data_size)]
    df = spark.createDataFrame(data, ["number"])

    start_time = time.time()
    # Hisoblash: har bir raqamni kvadratga ko'tarish
    result = df.select((col("number") ** 2).alias("squared"))
    sum_result = result.agg(spark_sum("squared")).collect()[0][0]
    end_time = time.time()

    print(f"PySpark vaqti: {end_time - start_time:.4f} soniya")
    spark.stop()  # SparkSession'ni to'xtatish
    return sum_result
```

```python
data_size = 1000000  # 1 million qator

pandas_result = pandas_example(data_size)
pyspark_result = pyspark_example(data_size)

print(f"Pandas natijasi: {pandas_result}")
print(f"PySpark natijasi: {pyspark_result}")
```

```python
# Katta hajmdagi ma'lumotlar bilan ishlash
print("\nKatta hajmdagi ma'lumotlar bilan test:")
data_size_large = 10000000  # 10 million qator
pandas_result_large = pandas_example(data_size_large)
pyspark_result_large = pyspark_example(data_size_large)

print(f"Pandas natijasi (katta): {pandas_result_large}")
print(f"PySpark natijasi (katta): {pyspark_result_large}")
```
